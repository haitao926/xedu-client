目标：理解图像数字化，用 Python 读取和操作图像。

提示：请先准备一张图片，并把代码中的 `image_path` 改成实际路径。


```python
# 第1步：导入PIL库中的Image类，用来读取和处理图片
from PIL import Image
```

```python
# 第2步：设置示例图片路径并读取图片（请改成你自己的图片路径）
image_path = 'images/sample_building.jpg'
img = Image.open(image_path).convert('RGB')

# 显示读取到的图片
img
```

```python
# 第3步：查看图片尺寸、颜色模式和像素总数
width, height = img.size
print('图片尺寸：', img.size)
print('宽度：', width, '像素')
print('高度：', height, '像素')
print('颜色模式：', img.mode)
print('像素总数：', width * height)
```

```python
# 第4步：读取特定位置的像素值（RGB）
x, y = width // 3, height // 3  # 选一个靠中间的位置
r, g, b = img.getpixel((x, y))
print(f'坐标 ({x}, {y}) 的像素值是 RGB = ({r}, {g}, {b})')
```

```python
# 第5步：复制图片，并把中心区域改成红色
img_red = img.copy()

# 计算中心区域（宽和高各取20%）
cx, cy = width // 2, height // 2
half_w = width // 10
half_h = height // 10

# 防止越界，把坐标限制在图片范围内
x_start = max(0, cx - half_w)
x_end = min(width, cx + half_w)
y_start = max(0, cy - half_h)
y_end = min(height, cy + half_h)

# 双层循环逐个修改像素颜色为红色 (255, 0, 0)
for px in range(x_start, x_end):
    for py in range(y_start, y_end):
        img_red.putpixel((px, py), (255, 0, 0))

# 显示修改后的图片
img_red
```

```python
# 第6步：保存修改后的图片
output_path = 'images/sample_building_red_center.jpg'
img_red.save(output_path)
print('保存完成：', output_path)
```
